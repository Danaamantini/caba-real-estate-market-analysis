```python
import pandas as pd

df = pd.read_csv("../data/raw/ar_properties.csv")
```

```python
caba_sales = df[
    (df["operation_type"] == "Venta") &
    (df["property_type"] == "Departamento") &
    (df["l2"] == "Capital Federal") &
    (df["currency"] == "USD")
].copy()
```

```python
q1 = caba_sales["surface_total"].quantile(0.25)
q3 = caba_sales["surface_total"].quantile(0.75)
iqr = q3 - q1

print(q1)
print(q3)
print(iqr)
```

```python
lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

print(lower_limit)
print(upper_limit)
```

```python
(caba_sales["surface_total"] > upper_limit).sum()
```

```python
caba_sales["price"].isna().sum()
```

```python
(caba_sales["price"] <= 0).sum()
```

```python
caba_sales["price"].describe()
```

```python
caba_sales[caba_sales["price"] < 20_000][
    ["price", "surface_total", "rooms", "bedrooms", "l3"]
]
```

```python
q1_price = caba_sales["price"].quantile(0.25)
q3_price = caba_sales["price"].quantile(0.75)

iqr_price = q3_price - q1_price

print(q1_price)
print(q3_price)
print(iqr_price)
```

```python
lower_limit_price = q1_price - 1.5 * iqr_price
upper_limit_price = q3_price + 1.5 * iqr_price

print(lower_limit_price)
print(upper_limit_price)
```

```python
(caba_sales["price"] > upper_limit_price).sum()
```

```python
caba_sales[caba_sales["price"] > 5_000_000][
    ["price", "surface_total", "rooms", "bedrooms", "l3"]
]
```

```python
valid_surface = caba_sales[
    caba_sales["surface_total"] > 0
].copy()

print(valid_surface.shape)
```

```python
valid_surface["price_per_m2"] = (
    valid_surface["price"] / valid_surface["surface_total"]
)

valid_surface["price_per_m2"].describe()
```

```python
valid_surface[
    valid_surface["price_per_m2"] > 20_000
][
    ["price", "surface_total", "price_per_m2", "rooms", "l3"]
].sort_values("price_per_m2", ascending=False)
```

```python
valid_surface[
    valid_surface["price_per_m2"] > 20_000
][
    ["price", "surface_total", "surface_covered", "price_per_m2", "rooms", "l3"]
].sort_values("price_per_m2", ascending=False)
```

```python
(
    valid_surface["surface_covered"] > valid_surface["surface_total"]
).sum()
```

```python
valid_surface["surface_difference"] = (
    valid_surface["surface_covered"] - valid_surface["surface_total"]
)
```

```python
valid_surface[
    valid_surface["surface_covered"] > valid_surface["surface_total"]
][
    ["surface_total", "surface_covered", "surface_difference", "price", "rooms", "l3"]
].sort_values("surface_difference", ascending=False)
```

```python
valid_surface[
    valid_surface["surface_difference"] > 0
]["surface_difference"].describe()
```

```python
(valid_surface["surface_difference"] > 50).sum()
```

```python
analysis_df = caba_sales[
    caba_sales["surface_total"].notna() &
    (caba_sales["surface_total"] > 0) &
    (caba_sales["price"] > 0)
].copy()
```

```python
analysis_df["price_per_m2"] = (
    analysis_df["price"] / analysis_df["surface_total"]
)
```

```python
print(analysis_df.shape)
analysis_df[
    ["price", "surface_total", "surface_covered", "price_per_m2"]
].describe()
```

```python
analysis_df.to_csv(
    "../data/processed/caba_apartments_clean.csv",
    index=False
)
```
